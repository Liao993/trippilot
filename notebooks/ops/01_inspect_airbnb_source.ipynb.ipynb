```python
from pyspark.sql import functions as F

RAW_PATH = (
    "/Volumes/trippilot_dev/bronze/source_files/"
    "insideairbnb/toronto/listings.csv.gz"
)

CITY = "toronto"
SOURCE = "insideairbnb"
FILE_FORMAT = "csv"
```

```python
df = (
    spark.read
    .option("header", True)
    .option("inferSchema", False)
    .option("multiLine", True)
    .option("escape", '"')
    .csv(RAW_PATH)
)

display(df.limit(20))
```

```python
print(f"City: {CITY}")
print(f"Source: {SOURCE}")
print(f"Raw path: {RAW_PATH}")
print(f"Row count: {df.count()}")

df.printSchema()
```

```python
identifier = "id"

id_summary = df.select(
    F.count("*").alias("total_rows"),
    F.countDistinct(F.col(identifier)).alias("distinct_ids"),
    F.count(F.when(F.col(identifier).isNull(), 1)).alias("null_ids"),
    F.count(F.when(F.trim(F.col(identifier)) == "", 1)).alias("blank_ids"),
)

display(id_summary)
```

```python
important_columns = [
    "id",
    "listing_url",
    "name",
    "host_id",
    "host_name",
    "neighbourhood_cleansed",
    "latitude",
    "longitude",
    "room_type",
    "price",
    "minimum_nights",
    "availability_365",
    "last_review",
]

existing_columns = [c for c in important_columns if c in df.columns]

display(
    df.select(existing_columns).limit(20)
)
```
